# Teen Mental Health Dataset - Exploratory Data Analysis (EDA)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
```

## Load Dataset

```python
df = pd.read_csv('../data/Teen_Mental_Health_Dataset.csv')
print('Shape:', df.shape)
df.head()
```

## Dataset Overview

```python
df.info()
df.describe()
```

## Missing Values Check

```python
df.isnull().sum()
```

## Class Distribution

```python
plt.figure(figsize=(6,4))
sns.countplot(x='depression_label', data=df)
plt.title('Depression Distribution')
plt.show()

print(df['depression_label'].value_counts())
```

## Correlation Heatmap

```python
df_corr = df.copy()
for col in df_corr.select_dtypes(include='object').columns:
    df_corr[col] = pd.factorize(df_corr[col])[0]

plt.figure(figsize=(10,6))
sns.heatmap(df_corr.corr(), cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## Histograms

```python
df.hist(figsize=(12,8), bins=20)
plt.tight_layout()
plt.show()
```

## Anxiety Level vs Depression

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='depression_label', y='anxiety_level', data=df)
plt.title('Anxiety Level vs Depression')
plt.show()
```

# Key Findings

- Dataset contains 1200 records and 13 features.
- No missing values were detected.
- Depression cases are significantly fewer than non-depression cases.
- Anxiety and stress levels are likely important predictors.
- Sleep and social media usage may influence depression risk.
- Random Forest model achieved strong performance on this dataset.
